```python
with open(catalog_dir + "SMBH_mass_catalog.txt") as f:
    for i in range(15):
        print(f"{i}: {f.readline().rstrip()}")
```

```python
import pandas as pd
from astropy.table import Table

path = catalog_dir + "SMBH_mass_catalog.txt"

# Define column specs as (start, end) pairs:
colspecs = [
    (0, 17),   # Fermi_Name
    (18, 20),  # Class
    (21, 28),  # Redshift
    (29, 38),  # Photon_Flux
    (39, 47),  # Photon_Flux_Unc
    (48, 52),  # Gamma
    (53, 59),  # logLha
    (60, 65),  # elogL_ha
    (66, 72),  # logLHb
    (73, 78), # e_logLHb
    (79, 86),# logLMgII
    (87, 92),# e_logLMgII
    (93, 98),# logLCIV
    (99, 104),# e_logLCIV
    (105, 110),# logMBH
    (111, 116) # e_logMBH
]

names = [
    'Fermi_Name','Class','Redshift','Photon_Flux','Photon_Flux_Unc',
    'Gamma','logLha','elogL_ha','logLHb','e_logLHb',
    'logLMgII','e_logLMgII','logLCIV','e_logLCIV',
    'logMBH','e_logMBH'
]

# Read with fixed widths
df = pd.read_fwf(path, colspecs=colspecs, names=names)

# Convert empty strings to None
df = df.replace(r'^\s*$', None, regex=True)

# Convert numeric columns to floats
for col in names[2:]:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Convert to Astropy Table
Black_Hole_Mass_Catalog = Table.from_pandas(df)
Black_Hole_Mass_Catalog.pprint_all()
```

```python
Black_Hole_Mass_Catalog.write(catalog_dir + 'Black_Hole_Mass_Catalog.fits',format='fits', overwrite = True)
```
