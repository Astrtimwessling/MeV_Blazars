```python
def mask_catalog_by_class(catalog, class_column):
    fsrq_mask = [x in ['FSRQ','fsrq'] for x in catalog[class_column].tolist()]
    bll_mask = [x in ['BLL','bll'] for x in catalog[class_column].tolist()]
    bcu_mask = [x in ['BCU','bcu'] for x in catalog[class_column].tolist()]

    fsrq_masked_catalog = catalog[fsrq_mask]
    bll_masked_catalog = catalog[bll_mask]
    bcu_masked_catalog = catalog[bcu_mask]

    return fsrq_masked_catalog, bll_masked_catalog, bcu_masked_catalog

def line(x_data,m,c): 
    return m*x_data + c 

c = 2.99792458e10         # speed of light in cm/s
H_0 = 70 * 1e5 / 3.085677581e24   # H0 = 70 km/s/Mpc → s^-1
omega_m = 0.3

def Luminosity(z, flux, flux_err):
    def f(x):
        return 1.0 / np.sqrt(omega_m *(1 + x)**3 + (1 - omega_m))
    
    integral, _ = quad(f, 1, 1 + z)
    D_c = (c / H_0) * integral
    d_L = (1 + z) * D_c

    L = 4 * math.pi * d_L**2 * flux
    L_err = L * (flux_err / flux)
    return L, L_err

def Asymmetric_Luminosity(z, flux, flux_err_low, flux_err_high):

    def Ez_inv(x):
        return 1.0 / np.sqrt(omega_m * x**3 + (1 - omega_m))

    # --- Comoving distance ---
    integral, _ = quad(Ez_inv, 1, 1 + z)
    D_c = (c / H_0) * integral      # cm

    # --- Luminosity distance ---
    d_L = (1 + z) * D_c             # cm

    # --- Luminosity ---
    L = 4 * math.pi * d_L**2 * flux

    # --- Asymmetric luminosity errors ---
    L_low = 4 * math.pi * d_L**2 * (flux - flux_err_low)
    L_high = 4 * math.pi * d_L**2 * (flux + flux_err_high)

    L_err_low = L - L_low
    L_err_high = L_high - L

    return L, L_err_low, L_err_high
```

```python
def dnde_PL(E, K, E0, gamma):
    return K * (E / E0) ** (-gamma)

def dnde_LP(E, K, E0, alpha, beta):
    return K * (E / E0) ** (-alpha - beta * np.log(E / E0))

def dnde_PLexpcut(E, K, E0, gamma, a, b):
    return K * (E / E0) ** (-gamma) * np.exp(a*((E0 ** b) - (E ** b)))
```
